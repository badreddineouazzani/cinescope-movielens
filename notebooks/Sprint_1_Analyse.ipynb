# Sprint 1 · Analyse exploratoire

MovieLens 100K, préparation et classement de référence. [Source GroupLens](https://grouplens.org/datasets/movielens/100k/).

```python
from pathlib import Path
import sys

project = Path.cwd()
if not (project / "cinescope").is_dir():
    project = project.parent
sys.path.insert(0, str(project))

import matplotlib.pyplot as plt
from cinescope.data import load
from cinescope.recommender import popularity

movies, ratings = load()
print(f"{len(movies):,} films, {len(ratings):,} notes, {ratings.user_id.nunique():,} utilisateurs")
ratings["rating"].describe()
```

## Distribution des notes

```python
counts = ratings["rating"].value_counts().sort_index()
counts.plot.bar(color="#662E91", rot=0, xlabel="Note sur 5", ylabel="Nombre de notes")
plt.tight_layout()
plt.show()
```

## Activité et films les plus notés

```python
print("Notes par utilisateur")
display(ratings.groupby("user_id").size().describe())
votes = ratings["movie_id"].value_counts().head(10)
labels = movies.set_index("movie_id").loc[votes.index, "title"]
votes.set_axis(labels).sort_values().plot.barh(
    color="#662E91", xlabel="Nombre de notes", ylabel=""
)
plt.tight_layout()
plt.show()
```

## Top 10 des films les mieux notés

Référence de classement avec un seuil de 20 votes.

```python
top = popularity(ratings, min_votes=20).head(10).join(movies.set_index("movie_id")["title"])
display(top[["title", "average", "votes"]].round(2))
assert len(top) == 10 and (top["votes"] >= 20).all()
```
